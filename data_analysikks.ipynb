```

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import plotly.express as px
import plotly.graph_objects as go
    
```

## Загрузка данных

```
# Загрузка набора данных: Загрузка CSV 7bc5
df_e89101bd = pd.read_csv('data/customers.csv', sep=',')
print(f'Размер загруженного датафрейма: {df_e89101bd.shape}')
```

## Визуализации

```
# Базовые визуализации для df_e89101bd

# Гистограммы для числовых колонок
numeric_cols = df_e89101bd.select_dtypes(include=['number']).columns
for col in numeric_cols[:5]:  # Ограничение первыми 5 числовыми колонками
    fig = px.histogram(df_e89101bd, x=col, title=f'Распределение {{col}}')
    fig.show()

# Столбчатые диаграммы для категориальных колонок
cat_cols = df_e89101bd.select_dtypes(include=['object', 'category']).columns
for col in cat_cols[:5]:  # Ограничение первыми 5 категориальными колонками
    value_counts = df_e89101bd[col].value_counts().reset_index()
    value_counts.columns = [col, 'count']
    fig = px.bar(value_counts, x=col, y='count', title=f'Количество {{col}}')
    fig.show()
```
